# <span style="color:#4F46E5; font-weight:700;">1- Binary Search</span>

Binary search is a way to find a value in a sorted list much faster than checking every item one by one.

Instead of scanning the whole list, we repeatedly split the list in half and focus on the side where the value could be.

```python
def binary_search(arr, target):
    """
    Perform binary search on a sorted array to find the index of the target value.

    Parameters:
    arr (list): A sorted list of elements.
    target: The value to search for in the array.

    Returns:
    int: The index of the target value if found, otherwise -1.
    """
    left, right = 0, len(arr) - 1

    while left <= right:
        mid = left + (right - left) // 2

        # Check if the target is present at mid
        if arr[mid] == target:
            return mid
        # If target is greater, ignore left half
        elif arr[mid] < target:
            left = mid + 1
        # If target is smaller, ignore right half
        else:
            right = mid - 1

    # Target was not found in the array
    return -1
```

### Why it matters

A normal search checks every item, which is O(n). Binary search reduces that to O(log n), which is far faster for large datasets.

# <span style="color:#4F46E5; font-weight:700;">2- Big O Notation</span>


## What is Big O?

Big O notation describes how the runtime or memory usage of an algorithm grows as the input size increases.

It helps us compare algorithms without focusing on the exact speed of a machine.

- **O(1):** Constant time
- **O(log n):** Logarithmic time
- **O(n):** Linear time
- **O(n log n):** Often seen in efficient sorting
- **O(n²):** Quadratic time

Big O helps us compare algorithms by focusing on their growth rate rather than exact execution time. It is useful for understanding how scalable an algorithm is when input size becomes large.

**Example:** Binary search runs in **O(log n)**, while a simple loop over every item runs in **O(n)**.



### Why it matters

When the input becomes very large, the difference between O(n) and O(log n) becomes massive. This is the heart of algorithm design.

# <span style="color:#4F46E5; font-weight:700;">4- Selection Sort</span>

Selection sort repeatedly finds the smallest remaining element and moves it to the front.

This is simple to understand, but not the fastest algorithm for large lists.

```python
def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_index = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_index]:
                min_index = j
        arr[i], arr[min_index] = arr[min_index], arr[i]
    return arr

print(selection_sort([5, 3, 8, 1, 9]))
# [1, 3, 5, 8, 9]
```


### Why it matters

Selection sort is easy to reason about and is useful for learning the idea of repeated comparisons. Its cost is O(n²), which is not efficient for large inputs.

# <span style="color:#4F46E5; font-weight:700;">5- Recursion</span>

Recursion happens when a function calls itself. It is a powerful pattern for problems that can be broken into smaller, similar versions of the same problem.

```python
def factorial(n):
    return 1 if n == 1 else n * factorial(n - 1)

print(factorial(5))  # 120

# This works because the function keeps calling itself with a smaller input until it reaches the base case.
```

### Why it matters

Recursion is often used in tree traversal, divide-and-conquer algorithms, and problems that naturally split into subproblems.

# <span style="color:#4F46E5; font-weight:700;">6- Quicksort</span>


Quicksort is a divide-and-conquer algorithm. It picks a pivot, places smaller values left and larger values right, and then sorts the smaller partitions recursively.

```python
def quicksort(arr):
    if len(arr) < 2:
        return arr

    pivot = arr[0]
    less = [i for i in arr[1:] if i <= pivot]
    greater = [i for i in arr[1:] if i > pivot]

    return quicksort(less) + [pivot] + quicksort(greater)

print(quicksort([10, 5, 2, 3, 7]))
# [2, 3, 5, 7, 10]
```

### Why it matters

Quicksort is much faster than selection sort in many real situations. Its average runtime is O(n log n), though worst-case performance can degrade to O(n²) if the pivot is poorly chosen.

# <span style="color:#4F46E5; font-weight:700;">8- Breadth-First Search (BFS)</span>


Breadth-first search explores a graph level by level. It is perfect for finding the shortest path in an unweighted graph.


```python
# Think of a social network or a map where each friend is connected to other friends.
from collections import deque

graph = {
    "You": ["Alice", "Bob", "Claire"],
    "Alice": ["Peggy"],
    "Bob": ["Anuj", "Peanut"],
    "Claire": ["Thom", "Jonny"],
    "Peggy": [],
    "Anuj": [],
    "Peanut": [],
    "Thom": [],
    "Jonny": []
}


def bfs(graph, start):
    visited = {start}
    queue = deque([start])

    while queue:
        person = queue.popleft()
        print(person)

        for friend in graph[person]:
            if friend not in visited:
                visited.add(friend)
                queue.append(friend)

bfs(graph, "You")
```

### Why it matters

BFS is useful when you need to find the shortest number of steps between places, people, or nodes in a graph.

# <span style="color:#4F46E5; font-weight:700;">9- Dijkstra's Algorithm</span>


Dijkstra's algorithm finds the shortest path in a weighted graph. Unlike BFS, it takes edge costs into account.

```python
# Imagine a map with roads of different lengths. Some roads are short, some are longer, and we want the cheapest route.
import heapq

graph = {
    "Start": {"A": 6, "B": 2},
    "A": {"Finish": 1},
    "B": {"A": 3, "Finish": 5},
    "Finish": {}
}


def dijkstra(graph, start):
    distances = {node: float('inf') for node in graph}
    distances[start] = 0
    heap = [(0, start)]

    while heap:
        current_distance, node = heapq.heappop(heap)

        if current_distance > distances[node]:
            continue

        for neighbor, weight in graph[node].items():
            new_distance = current_distance + weight
            if new_distance < distances[neighbor]:
                distances[neighbor] = new_distance
                heapq.heappush(heap, (new_distance, neighbor))

    return distances

print(dijkstra(graph, "Start"))
```

### Why it matters

Dijkstra's algorithm is used in GPS systems, network routing, and many optimization problems where path cost matters.

# <span style="color:#4F46E5; font-weight:700;">10- Greedy Algorithms</span>


A greedy algorithm makes the best choice at each step, hoping that local optimal decisions will lead to the global optimum.

```python
# Suppose you have coin denominations and want to make the amount 36 using the fewest coins.
coins = [1, 2, 5, 10, 20]

def greedy_change(amount, coin_values):
    result = []
    for coin in sorted(coin_values, reverse=True):
        while amount >= coin:
            result.append(coin)
            amount -= coin
    return result

print(greedy_change(36, coins))
# [20, 10, 5, 1]
```

### Why it matters

Greedy algorithms are fast and simple, but they only work when the greedy choice is safe for the problem. They are often used in scheduling and optimization tasks.

# <span style="color:#4F46E5; font-weight:700;">11- Dynamic Programming</span>


Dynamic programming solves a problem by breaking it into overlapping subproblems and remembering the answers to avoid recomputation.

```python
# The Fibonacci sequence is a classic example.
def fibonacci(n, memo={}):  # sourcery skip: default-mutable-arg
    if n in memo:
        return memo[n]
    if n <= 1:
        return n
    memo[n] = fibonacci(n - 1, memo) + fibonacci(n - 2, memo)
    return memo[n]

print(fibonacci(10))  # 55
```

### Why it matters

Dynamic programming is used when a problem can be solved by combining the results of smaller versions of itself. It is a key technique for optimization and planning problems.
